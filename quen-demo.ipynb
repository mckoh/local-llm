```python
import numpy as np
import ollama

# 1. Modell definieren (Beispiel mit Qwen)
MODEL_NAME = "qwen3-embedding"  # Oder dein spezifisches Embedding-Modell

# 2. Zwei semantisch ähnliche Sätze und einen unpassenden Satz definieren
text1 = "Der Hund bellt im Garten."
text2 = "Ein Vierbeiner macht Lärm draußen."
text3 = "Python ist eine Programmiersprache."

# 3. Embeddings über Ollama generieren
response1 = ollama.embeddings(model=MODEL_NAME, prompt=text1)
response2 = ollama.embeddings(model=MODEL_NAME, prompt=text2)
response3 = ollama.embeddings(model=MODEL_NAME, prompt=text3)

vec1 = np.array(response1.embedding)
vec2 = np.array(response2.embedding)
vec3 = np.array(response3.embedding)
```

```python
def cosine_similarity(a, b):
    return round(np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b)), 2)

print("Ähnlichkeit von Text 1 und Text 2", cosine_similarity(vec1, vec2))
print("Ähnlichkeit von Text 2 und Text 3", cosine_similarity(vec2, vec3))
```
